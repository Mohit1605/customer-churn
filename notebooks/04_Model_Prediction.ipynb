```python
import warnings
warnings.filterwarnings("ignore")

import joblib
import pandas as pd
import numpy as np
```

```python
pipeline = joblib.load(
    "/content/drive/MyDrive/Customer Churn/customer_churn_pipeline.pkl"
)

print(type(pipeline))
```

```python
def create_features(df):

    internet_services = [
        "OnlineSecurity",
        "OnlineBackup",
        "DeviceProtection",
        "TechSupport",
        "StreamingTV",
        "StreamingMovies"
    ]

    df = df.copy()

    df["ServiceCount"] = (
        df[internet_services]
        .eq("Yes")
        .sum(axis=1)
    )

    df["IsNewCustomer"] = (
        df["tenure"] <= 12
    ).astype(int)

    return df
```

```python
def predict_customer(data):

    df = pd.DataFrame([data])

    df = create_features(df)

    probability = pipeline.predict_proba(df)[0][1]

    prediction = pipeline.predict(df)[0]

    label = "Yes" if prediction == 1 else "No"

    return {
        "Prediction": label,
        "Churn Probability": round(probability * 100,2)
    }
```

```python
customer = {

    "gender":"Female",

    "SeniorCitizen":0,

    "Partner":"Yes",

    "Dependents":"No",

    "tenure":8,

    "PhoneService":"Yes",

    "MultipleLines":"No",

    "InternetService":"Fiber optic",

    "OnlineSecurity":"No",

    "OnlineBackup":"Yes",

    "DeviceProtection":"No",

    "TechSupport":"No",

    "StreamingTV":"Yes",

    "StreamingMovies":"Yes",

    "Contract":"Month-to-month",

    "PaperlessBilling":"Yes",

    "PaymentMethod":"Electronic check",

    "MonthlyCharges":92.5,

    "TotalCharges":740.0

}
```

```python
predict_customer(customer)
```
